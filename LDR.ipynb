```python
import numpy as np
import matplotlib.pyplot as plt

distance = np.array([5,10,15,20,25,30,35,40])

# Simulated values
simulated = 1000*(distance/5)**2

# Example experimental values
experimental = np.array([980,4050,9050,15900,24800,35600,48700,64500])

plt.plot(distance, simulated, 'b-', label='Simulation')
plt.scatter(distance, experimental, color='red', label='Experimental')

plt.xlabel("Distance (cm)")
plt.ylabel("Resistance (Ω)")
plt.title("LDR Resistance vs Distance")
plt.legend()
plt.grid(True)
plt.show()
```
